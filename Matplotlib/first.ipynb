```python
import matplotlib.pyplot as plt
import numpy as np

x = ['c++', 'python' , 'java', 'ML']
y = [89,95,45,60]
z = [45,32,40,100]

plt.xlabel("Language")
plt.ylabel("Learning")
plt.title("Overall")

plt.bar(x,y, width=0.6, edgecolor="red",label="ok1")
plt.bar(x,z ,width=0.6,edgecolor="red",label="ok2")

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = ['c++', 'python' , 'java', 'ML']
y = [89,95,45,60]
z = [45,32,40,100]

width=0.2

p = np.arange(len(x))
p1 = [j+width for j in p]

plt.xlabel("Language")
plt.ylabel("Learning")
plt.title("Overall")

plt.bar(p,y, width, label="ok1")
plt.bar(p1,z, width,label="ok2")
plt.xticks(p+width/2, x, rotation = 25)

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = ['c++', 'python' , 'java', 'ML']
y = [89,95,45,60]
z = [45,32,40,100]

plt.xlabel("Language")
plt.ylabel("Learning")
plt.title("Overall")

plt.barh(x,y ,label="ok1")
plt.barh(x,z ,label="ok2")

plt.legend()
plt.show()
```
